```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import (MultipleLocator, AutoMinorLocator)
import matplotlib.gridspec as gridspec
```

```python
plt.rcParams['figure.dpi'] = 120
plt.rcParams['font.family'] = 'Nimbus Sans'
plt.rcParams['font.size'] = '15'
```

```python
common_path='/nfs/roberts/home/ys668/analysis/run12/data/fully_corrected/'
```

```python
# mean_20_25=np.loadtxt(common_path+'txts/m/20_25_mean.txt',delimiter=',')
# err_20_25 =np.loadtxt(common_path+'txts/m/20_25_err.txt',delimiter=',')
# unf_20_25 =np.loadtxt(common_path+'txts/m/20_25_unf.txt',delimiter=',')
mean_20_25_isaac =np.loadtxt(common_path+'txts/m/20_25_mean_isaac.txt',delimiter=',')
err_20_25_isaac = np.loadtxt(common_path+'txts/m/20_25_err_isaac.txt',delimiter=',')
unf_20_25_isaac = np.loadtxt(common_path+'txts/m/20_25_unf_isaac.txt',delimiter=',')*mean_20_25_isaac
```

```python
# mean_25_30=np.loadtxt(common_path+'txts/m/25_30_mean.txt',delimiter=',')
# err_25_30 =np.loadtxt(common_path+'txts/m/25_30_err.txt',delimiter=',')
# unf_25_30 =np.loadtxt(common_path+'txts/m/25_30_unf.txt',delimiter=',')
mean_25_30_isaac =np.loadtxt(common_path+'txts/m/25_30_mean_isaac.txt',delimiter=',')
err_25_30_isaac = np.loadtxt(common_path+'txts/m/25_30_err_isaac.txt',delimiter=',')
unf_25_30_isaac = np.loadtxt(common_path+'txts/m/25_30_unf_isaac.txt',delimiter=',')*mean_25_30_isaac
```

```python
# mean_30_40=np.loadtxt(common_path+'txts/m/30_40_mean.txt',delimiter=',')
# err_30_40 =np.loadtxt(common_path+'txts/m/30_40_err.txt',delimiter=',')
# unf_30_40 =np.loadtxt(common_path+'txts/m/30_40_unf.txt',delimiter=',')
mean_30_40_isaac =np.loadtxt(common_path+'txts/m/30_40_mean_isaac.txt',delimiter=',')
err_30_40_isaac = np.loadtxt(common_path+'txts/m/30_40_err_isaac.txt',delimiter=',')
unf_30_40_isaac = np.loadtxt(common_path+'txts/m/30_40_unf_isaac.txt',delimiter=',')*mean_30_40_isaac
```

```python
unf_tot_20_25=np.sqrt(unf_20_25**2+unf_20_25_isaac**2)
unf_tot_25_30=np.sqrt(unf_25_30**2+unf_25_30_isaac**2)
unf_tot_30_40=np.sqrt(unf_30_40**2+unf_30_40_isaac**2)
```

```python
gs1 = gridspec.GridSpec(2,3,width_ratios=(1,1,1), height_ratios=(8,3))

fig = plt.figure(figsize=(12,5))
plt.subplots_adjust(wspace=0, hspace=0)

x = np.arange(0.5,12.5)

ax0 = fig.add_subplot(gs1[0])
plt.text(5.5,0.27, 'STAR Preliminary', color='r', fontstyle='italic')
plt.text(6.75,0.175,'$20 < p_\mathrm{T} < 25$ GeV/$c$', fontsize=10)
plt.fill_between(x, mean_20_25-err_20_25, mean_20_25+err_20_25, color='r', alpha=0.15)
plt.fill_between(x+0, mean_20_25_isaac-err_20_25_isaac, mean_20_25_isaac+err_20_25_isaac, color='k', alpha=0.15)
plt.plot(x, mean_20_25, color='red', marker='*', ls='none', markersize=10)
plt.plot(x+0, mean_20_25_isaac, label='RooUnfold\nPRD (2021)', ls='none', marker='*', markersize=10, color='none', mec='k')
plt.xlim(left=0, right=12)
plt.ylabel('1/N dN/dM [$c^2$/GeV]')
plt.ylim(-0.01,0.3)
ax = plt.gca()
ax.tick_params(top=True, right=True, bottom=True, left=True, direction='in', which='both')
ax.axes.xaxis.set_ticklabels([])
ax.axes.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.axes.xaxis.set_major_locator(MultipleLocator(2))
ax.axes.yaxis.set_minor_locator(MultipleLocator(0.01))
ax.axes.yaxis.set_major_locator(MultipleLocator(0.05))

ax1 = fig.add_subplot(gs1[1])
plt.text(3.,0.25,'$p+p$  $\sqrt{s}=200$ GeV\nanti-$k_\mathrm{T}$ full jets, R=0.4, |$\eta$|<0.6', fontsize=12)
plt.text(6.75,0.175,'$25 < p_\mathrm{T} < 30$ GeV/$c$', fontsize=10)
plt.plot(x, mean_25_30, color='red', marker='*', ls='none', markersize=10, label='MultiFold')
plt.fill_between(x, mean_25_30-err_25_30, mean_25_30+err_25_30, color='r', alpha=0.15)
plt.fill_between(x+0, mean_25_30_isaac-err_25_30_isaac, mean_25_30_isaac+err_25_30_isaac, color='k', alpha=0.15)
plt.plot(x+0, mean_25_30_isaac, label='RooUnfold\nPRD (2021)', ls='none', marker='*', markersize=10, color='none', mec='k')
plt.xlim(left=0, right=12)
plt.ylim(-0.01,0.3)
ax = plt.gca()
ax.tick_params(top=True, right=True, bottom=True, left=True, direction='in', which='both')
ax.axes.yaxis.set_ticklabels([])
ax.axes.xaxis.set_ticklabels([])
ax.axes.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.axes.xaxis.set_major_locator(MultipleLocator(2))
ax.axes.yaxis.set_minor_locator(MultipleLocator(0.01))
ax.axes.yaxis.set_major_locator(MultipleLocator(0.05))

ax2 = fig.add_subplot(gs1[2])
plt.text(6.75,0.175,'$30 < p_\mathrm{T} < 40$ GeV/$c$', fontsize=10)
plt.plot(x, mean_30_40, color='red', marker='*', ls='none', markersize=10, label='MultiFold')
plt.fill_between(x, mean_30_40-err_30_40, mean_30_40+err_30_40, color='r', alpha=0.15)
plt.fill_between(x+0, mean_30_40_isaac-err_30_40_isaac, mean_30_40_isaac+err_30_40_isaac, color='k', alpha=0.15)
plt.plot(x+0, mean_30_40_isaac, label='RooUnfold\nPRD (2021)', ls='none', marker='*', markersize=10, color='none', mec='k')
plt.xlim(left=0, right=12)
plt.ylim(-0.01,0.3)
ax = plt.gca()
ax.tick_params(top=True, right=True, bottom=True, left=True, direction='in', which='both')
ax.axes.yaxis.set_ticklabels([])
ax.axes.xaxis.set_ticklabels([])
ax.axes.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.axes.xaxis.set_major_locator(MultipleLocator(2))
ax.axes.yaxis.set_minor_locator(MultipleLocator(0.01))
ax.axes.yaxis.set_major_locator(MultipleLocator(0.05))
plt.legend(frameon=False, fontsize=12, loc='upper right')

ax3 = fig.add_subplot(gs1[3])
plt.plot(x, mean_20_25/mean_20_25_isaac, color='red', marker='*', ls='none', markersize=10)
plt.fill_between(x, (mean_20_25_isaac-unf_tot_20_25)/mean_20_25_isaac, (mean_20_25_isaac+unf_tot_20_25)/mean_20_25_isaac, color='r', alpha=0.15)
plt.hlines(y=1.0, xmin=0, xmax=12, linewidth=0.5, linestyle='--', color='k')
plt.xlim(left=0, right=12)
plt.xlabel('M [GeV/$c^2$]')
plt.ylabel('Ratio with\nRooUnfold', fontsize=12)
plt.ylim(0.5,1.5)
plt.yticks([0.5,1.0],['0.5','1.0'], fontsize=12)
ax = plt.gca()
ax.tick_params(top=True, right=True, bottom=True, left=True, direction='in', which='both')
ax.axes.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.axes.xaxis.set_major_locator(MultipleLocator(2))
ax.axes.yaxis.set_minor_locator(MultipleLocator(0.1))
ax.axes.yaxis.set_major_locator(MultipleLocator(0.5))
plt.xticks(np.arange(0,12,2))

ax4 = fig.add_subplot(gs1[4])
plt.plot(x, mean_25_30/mean_25_30_isaac, color='red', marker='*', ls='none', markersize=10)
plt.fill_between(x, (mean_25_30_isaac-unf_tot_25_30)/mean_25_30_isaac, (mean_25_30_isaac+unf_tot_25_30)/mean_25_30_isaac, color='r', alpha=0.15)
plt.hlines(y=1.0, xmin=0, xmax=12, linewidth=0.5, linestyle='--', color='k')
plt.xlim(left=0, right=12)
plt.xlabel('M [GeV/$c^2$]')
plt.ylim(0.5,1.5)
ax = plt.gca()
ax.tick_params(top=True, right=True, bottom=True, left=True, direction='in', which='both')
ax.axes.yaxis.set_ticklabels([])
ax.axes.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.axes.xaxis.set_major_locator(MultipleLocator(2))
ax.axes.yaxis.set_minor_locator(MultipleLocator(0.1))
ax.axes.yaxis.set_major_locator(MultipleLocator(0.5))
plt.xticks(np.arange(0,12,2))

ax5 = fig.add_subplot(gs1[5])
plt.plot(x, mean_30_40/mean_30_40_isaac, color='red', marker='*', ls='none', markersize=10)
plt.fill_between(x, (mean_30_40_isaac-unf_tot_30_40)/mean_30_40_isaac, (mean_30_40_isaac+unf_tot_30_40)/mean_30_40_isaac, color='r', alpha=0.15, label='Uncorrelated sys. unc.')
plt.hlines(y=1.0, xmin=0, xmax=12, linewidth=0.5, linestyle='--', color='k')
plt.xlim(left=0, right=12)
plt.xticks(np.arange(0,14,2))
plt.xlabel('M [GeV/$c^2$]')
plt.ylim(0.5,1.5)
ax = plt.gca()
ax.tick_params(top=True, right=True, bottom=True, left=True, direction='in', which='both')
ax.axes.yaxis.set_ticklabels([])
ax.axes.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.axes.xaxis.set_major_locator(MultipleLocator(2))
ax.axes.yaxis.set_minor_locator(MultipleLocator(0.1))
ax.axes.yaxis.set_major_locator(MultipleLocator(0.5))
plt.legend(frameon=False, fontsize=10, loc='lower center')
```
